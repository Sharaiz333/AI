Creating a GPA calculator in python

```python
courses = int(input("Enter number of courses: "))

total_points = 0
total_credits = 0

for i in range(1, courses + 1):
    print(f"\nCourse {i}")
    grade_point = float(input("Enter grade point (e.g., 4.0, 3.5): "))
    credit_hours = int(input("Enter credit hours: "))

    total_points += grade_point * credit_hours
    total_credits += credit_hours

if total_credits == 0:
    print("\nError: Total credits cannot be zero.")
else:
    gpa = total_points / total_credits
print("Your GPA is:", round(gpa, 2))
```

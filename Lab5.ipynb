```python
# -------------------- DFS --------------------

def DFS(graph, start):
    stack = []
    visited = []

    stack.append(start)

    print("DFS Traversal:\n")

    while stack:
        node = stack.pop()

        if node not in visited:
            visited.append(node)

            print("Visited Node:", node)
            print("Adjacent Nodes:", graph[node])
            print("----------------------")


            for neighbour in reversed(graph[node]):
                if neighbour not in visited:
                    stack.append(neighbour)

graph = {
    0: [1, 2],
    1: [2, 3],
    2: [0, 4],
    3: [4],
    4: []
}

DFS(graph, 0)
```

```python
from collections import deque

def BFS(graph, start):
    queue = deque()
    visited = []

    queue.append(start)
    visited.append(start)

    print("BFS Traversal:\n")

    while queue:
        node = queue.popleft()

        print("Visited Node:", node)
        print("Adjacent Nodes:", graph[node])
        print("----------------------")

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.append(neighbour)
                queue.append(neighbour)

graph = {
    0: [1, 2],
    1: [2, 3],
    2: [0, 4],
    3: [4],
    4: []
}

BFS(graph, 0)
```

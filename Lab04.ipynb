# SETS IN PYTHON
# Create a Set(Using Curly Braces



```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
```

# Create a Set (Using Constructor)

```python
my_set = set([10, 20, 30, 40])
print(my_set)
```

# Add Element to Set


```python
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)
```

# Remove Element using remove()

```python
my_set = {1, 2, 3}
my_set.remove(2)
print(my_set)
```

# Remove Element using discard()

```python
my_set = {1, 2, 3}
my_set.discard(5)   # No error if element not found
print(my_set)
```

# Set Union

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

print(set1.union(set2))
# OR
print(set1 | set2)
```

# Set Intersection

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}

print(set1.intersection(set2))
# OR
print(set1 & set2)
```

# Set Difference

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}

print(set1.difference(set2))
# OR
print(set1 - set2)
```

# Check Item in Set

```python
my_set = {10, 20, 30}

print(20 in my_set)
print(40 in my_set)
```

# Remove Duplicates Using Set

```python
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_list = list(set(my_list))

print(unique_list)
```

# DICTIONARIES IN PYTHON
# Create Dictionary

```python
student = {
    "sapid": 12345,
    "name": "Ali",
    "age": 21,
    "semester": 6
}

print(student)
```

# Access Dictionary Items

```python
student = {"name": "Ali", "age": 21}

print(student["name"])
print(student.get("age"))
```

# Get All Keys

```python
student = {"name": "Ali", "age": 21}

print(student.keys())
```

# Change Dictionary Value

```python
student = {"name": "Ali", "age": 21}

student["age"] = 22
print(student)
```

# Update Dictionary

```python
student = {"name": "Ali", "age": 21}

student.update({"age": 23})
print(student)
```

# Remove Item using pop()

```python
student = {"name": "Ali", "age": 21}

student.pop("age")
print(student)
```

# Remove Last Item using popitem()

```python
student = {"name": "Ali", "age": 21}

student.popitem()
print(student)
```

# Remove Item using del

```python
student = {"name": "Ali", "age": 21}

del student["age"]
print(student)
```

# Clear Dictionary

```python
student = {"name": "Ali", "age": 21}

student.clear()
print(student)
```

# Task 1
# Create NumPy array, split into 3 arrays, calculate sum

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

split_arrays = np.array_split(arr, 3)

print("Split Arrays:", split_arrays)

total_sum = sum([np.sum(a) for a in split_arrays])

print("Total Sum:", total_sum)
```

```python
import numpy as np

arr = np.array([2, 6, 10, 14, 18, 20])

filtered = arr[(arr > 5) & (arr < 15)]

print(filtered)
```

# Task 3
# Create Sets and Perform Operations

```python
# Set1: Even numbers 0–10
set1 = {0, 2, 4, 6, 8, 10}

# Set2 using constructor: Positive integers 1–10
set2 = set(range(1, 11))

print("Set1:", set1)
print("Set2:", set2)

# Remove second and fifth element (convert to list first)
temp_list = list(set2)

temp_list.pop(1)   # remove second element
temp_list.pop(3)   # remove fifth element (index shifted)

set2 = set(temp_list)

print("Updated Set2:", set2)

# Intersection
set3 = set1.intersection(set2)

print("Intersection (Set3):", set3)
```

# Task 4
# Create Students Dictionary & Update Semester

```python
students = {
    1001: {"sapid": 1001, "name": "Ali", "age": 21, "semester": 6},
    1002: {"sapid": 1002, "name": "Ahmed", "age": 22, "semester": 6},
    1003: {"sapid": 1003, "name": "Sara", "age": 21, "semester": 6},
    1004: {"sapid": 1004, "name": "Ayesha", "age": 23, "semester": 6},
    1005: {"sapid": 1005, "name": "Bilal", "age": 22, "semester": 6}
}

# Update Semester from 6 to 7
for key in students:
    students[key]["semester"] = 7

print(students)
```

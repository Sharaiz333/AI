# Copy vs View
# Copy

```python
import numpy as np
arr = np.array([1,2,3])
x = arr.copy()
arr[0] = 10
print(arr)
print(x)
```

# View

```python
import numpy as np
arr = np.array([1,2,3])
x = arr.view()
arr[0] = 10
print(arr)
print(x)
```

# Array Iteration
# 1-D

```python
import numpy as np
arr = np.array([1,2,3])
for x in arr:
    print(x)
```

# 2-D

```python
import numpy as np
arr = np.array([[1,2],[3,4]])
for row in arr:
    for element in row:
        print(element)
```

# Using nditer()

```python
import numpy as np
arr = np.array([[1,2],[3,4]])
for x in np.nditer(arr):
    print(x)
```

# Joining Arrays

# concatenate()

```python
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
newarr = np.concatenate((arr1, arr2))
print(newarr)
```

# stack()

```python
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
newarr = np.stack((arr1, arr2), axis=1)
print(newarr)
```

# Splitting Arrays

```python
import numpy as np
arr = np.array([1,2,3,4,5,6])
newarr = np.array_split(arr, 3)
print(newarr)
```

# Searching Arrays

# where()

```python
import numpy as np
arr = np.array([4,5,6,7,8])
x = np.where(arr == 6)
print(x)
```

# searchsorted()

```python
import numpy as np
arr = np.array([1,3,5,7])
x = np.searchsorted(arr, 4)
print(x)
```

# Sorting Arrays

```python
import numpy as np
arr = np.array([3,1,2])
print(np.sort(arr))
```

# Filtering Arrays

```python
import numpy as np
arr = np.array([10, 15, 20, 25])
newarr = arr[arr > 15]
print(newarr)
```

# FINAL TASK

```python
import numpy as np

arr = np.arange(2, 11).reshape(3, 3)

arr = arr.astype(float)

flat = arr.reshape(-1)

print("Matrix:\n", arr)
print("Flattened:", flat)
print("Maximum value:", np.max(flat))
print("Minimum value:", np.min(flat))
```

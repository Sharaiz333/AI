# Confusion Matrix - 10 Examples (Binary Classification)

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report
)
```

## Example 1: Basic Manual Data

```python
y_true = [1, 0, 1, 1, 0]
y_pred = [1, 0, 0, 1, 0]

print("Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 2: Random Binary Data

```python
y_true = np.random.randint(0, 2, 50)
y_pred = np.random.randint(0, 2, 50)

print("Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 3: Logistic Regression

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=120, n_features=4, random_state=42)

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)

print("Accuracy:", accuracy_score(y, y_pred))
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 4: KNN

```python
from sklearn.datasets import make_classification
from sklearn.neighbors import KNeighborsClassifier

X, y = make_classification(n_samples=120, n_features=4, random_state=1)

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

y_pred = model.predict(X)

print("Accuracy:", accuracy_score(y, y_pred))
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 5: Decision Tree

```python
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier

X, y = make_classification(n_samples=120, n_features=4, random_state=2)

model = DecisionTreeClassifier()
model.fit(X, y)

y_pred = model.predict(X)

print("Accuracy:", accuracy_score(y, y_pred))
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 6: Train-Test Split

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=120, n_features=4, random_state=3)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 7: Imbalanced Dataset

```python
y_true = [0]*40 + [1]*10
y_pred = [0]*35 + [1]*15

print("Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 8: Iris Binary

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

data = load_iris()
X = data.data
y = (data.target == 0).astype(int)

model = LogisticRegression(max_iter=200)
model.fit(X, y)

y_pred = model.predict(X)

print("Accuracy:", accuracy_score(y, y_pred))
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 9: Threshold Adjustment

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=120, n_features=4, random_state=10)

model = LogisticRegression()
model.fit(X, y)

y_prob = model.predict_proba(X)[:, 1]
y_pred = (y_prob > 0.3).astype(int)

print("Accuracy:", accuracy_score(y, y_pred))
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)
ConfusionMatrixDisplay(cm).plot()
plt.show()
```

## Example 10: Custom Labels

```python
y_true = ["spam", "not spam", "spam", "spam", "not spam"]
y_pred = ["spam", "spam", "spam", "not spam", "not spam"]

print("Accuracy:", accuracy_score(y_true, y_pred))
print(classification_report(y_true, y_pred))

cm = confusion_matrix(y_true, y_pred, labels=["spam", "not spam"])
ConfusionMatrixDisplay(cm, display_labels=["spam", "not spam"]).plot()
plt.show()
```
